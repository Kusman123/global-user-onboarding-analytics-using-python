# Phase 3: Data Loading, Flattening & Initial Inspection

**Objective:** Load the raw JSON data fetched from the Random User API, flatten the deeply nested structure into a Pandas DataFrame, and perform an initial health check to understand data types, missing values, and basic statistics.

**Business Context:** Understanding the raw structure helps us decide what needs cleaning before we can build our "User Profile Quality Score".


```python
# Set the working directory to the project root
import os
import sys

# Get the absolute path of the current notebook's directory
notebook_dir = os.getcwd()
print(f"Current working directory: {notebook_dir}")

# Move up one level to the project root (since the notebook is inside 'notebooks/')
project_root = os.path.dirname(notebook_dir)
print(f"Project root directory: {project_root}")

# Change the working directory to the project root
os.chdir(project_root)

# Verify the change
print(f"New working directory: {os.getcwd()}")

if project_root not in sys.path:
    sys.path.insert(0, project_root)
    print("Project root added to Python path!")
```

```python
# Importing the libraries 
import numpy as np
import pandas as pd
import json
from datetime import datetime

print(f"Pandas version: {pd.__version__}")
```

```python
# Defining the path to our raw data folder
raw_data_folder = "data/raw/"

#  Geting a list of all files in the raw data folder
all_files = os.listdir(raw_data_folder)

# Filter to only include JSON files
json_files = [] # Step 1: Creating an empty list

for f in all_files: # Step 2: Look at every file name in all_files one by one
    if f.endswith(".json"): # Step 3: Check if the file name ends with '.json'
        json_files.append(f) # Step 4: If yes, add it to our new list
        
# Sorting them so the newest one is last, and take the last one
if json_files:
    latest_json_file = sorted(json_files)[-1]
else:
    print("No JSON files found!")

# Build the full file path
file_path = os.path.join(raw_data_folder, latest_json_file)

print(f"📂 Loading the most recent raw data file: {latest_json_file}")
print(f"📁 Full path: {file_path}")

# load the JSON file into a Python dictionary
with open(file_path,'r',encoding='utf-8') as f:
    raw_data = json.load(f)

# Check the type and the top-level keys
print(f"\n✅ Data loaded successfully!")
print(f"   Type of 'raw_data': {type(raw_data)}")
print(f"   Top-level keys in the JSON: {raw_data.keys()}")

```

```python
# The actual user data is inside the "results" key
users_list = raw_data['results']

# Let's check what we have
print(f"Total users fetched: {len(users_list)}")

# looking at the first user to understand the structure
first_user = users_list[0]
print(f"\nKeys inside a single user object: {first_user.keys()}")
```

```python
# flattening the nested dictionaries in to a flat table

df = pd.json_normalize(users_list,sep='_')

# Check the shape (rows, columns)
print(f"✅ DataFrame created successfully!")
print(f"   Shape (rows, columns): {df.shape}")
print(f"   Total columns: {df.shape[1]}")
print(f"\n📊 First 5 rows of the flattened DataFrame:")
df.head()
```

```python
print("===== DATAFRAME INFO =====")
df.info()

print("\n===== STATISTICAL SUMMARY (Numerical) =====")
df.describe()

# Check for missing values (nulls) in each column.
print("\n===== MISSING VALUES COUNT PER COLUMN =====")
missing_values = df.isnull().sum()

# Only show columns that have at least 1 missing value to keep it clean
missing_values[missing_values > 0].sort_values(ascending = False)
```

## Initial Observations & Business Insights (Business Notes)

### 1. Data Volume & Structure
- **Total Records:** 5,000 user profiles.
- **Total Features:** 34 columns after flattening.
- **Good Coverage:** We have a wide range of attributes, including personal details (`name`, `gender`), contact info (`email`, `phone`), location (`street`, `city`, `country`, `timezone`), login credentials, and `dob`/`registered` dates.

### 2. Data Quality & Missing Values (Critical Business Risk)
- **Missing Data:** Only one column has significant missing values:
  - `id_value` (National ID / Passport / SSN) has **931 missing values** (18.6% of users).
  - 
- **Business Impact:** 
  - This is a **major KYC (Know Your Customer) risk**. 
  - Nearly 1 in 5 users signed up without providing a valid government ID. 
  - These users **cannot be fully onboarded** and will require manual verification, costing the company time and money.
  - **Recommendation:** Flag these 931 users for immediate "Profile Completion" email campaigns.

### 3. Data Types & Feature Extraction Opportunities
- **Date Columns:** `dob_date` and `registered_date` are stored as strings (`object`/`str`).
  - *Action:* Must convert to `datetime` format to extract month, year, and day-of-week trends.
- **Age Columns (Pre-calculated):** `dob_age` (age now) and `registered_age` (years since registration) are integers. These are already clean.
- **Email:** Currently a plain string. We can extract `email_domain` (e.g., `@gmail.com`, `@yahoo.com`, `@protonmail.com`) to detect disposable/fraudulent emails.
- **Location:** We have `location_country` and `location_city`. Great for geographic segmentation.

---

### 4. Statistical Insights (Numerical Columns)
Looking at `df.describe()`:

- **`dob_age` (Age):**
  - Range: **25 to 81 years old**.
  - Mean: ~53 years old. 
  - **Insight:** The platform skews toward an **older demographic** (middle-aged to senior). Marketing campaigns should target this age group, and UX should be designed with accessibility in mind.

- **`registered_age` (Years since registration):**
  - Range: **4 to 24 years**.
  - Mean: ~14 years.
  - **Insight:** Most users have been on the platform for **over a decade** (average 14 years). This indicates **high brand loyalty and retention**. 
  - *Correlation note:* An 81-year-old user who registered 24 years ago would have been 57 at sign-up. A 25-year-old user who registered 4 years ago would have been 21. This data aligns perfectly.

- **`location_street_number`:**
  - Range: 1 to 9999. Mean is ~4,957.
  - **Insight:** Looks like random generation. No major outliers to worry about here.
